```python
import numpy as np
import pandas as pd

from sklearn.metrics import accuracy_score
```

```python
testPredPath = 'predictions/testPred.csv'
testPred = pd.read_csv(testPredPath)
testPred
```

```python
testLabels = np.load('data/testLabels.npy')
print(len(testLabels), set(testLabels))
testLabels
```

```python
classes = np.array(['Cytoplasm', 'Endoplasmic reticulum',
                    'Extracellular region', 'Mitochondria', 'Nucleus'])
classToLabel = {c:i for (i, c) in enumerate(classes)}
classToLabel
```

```python
preds = [classToLabel[c] for c in testPred['Class']] 
print(f'Accuracy: {accuracy_score(testLabels, preds)}')
```
